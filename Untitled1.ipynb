```python
from diffusers import StableDiffusionPipeline
import torch

# 1. Load the pre-trained Stable Diffusion model
model_id = "runwayml/stable-diffusion-v1-5"
pipe = StableDiffusionPipeline.from_pretrained(model_id, torch_dtype=torch.float16)
pipe = pipe.to("cuda") # Moves the model to the GPU for speed

# 2. Define your creative prompt
p = "A futuristic city in the style of Studio Ghibli, 8k resolution, cinematic lighting"

# 3. Generate the image
image = pipe(p).images[0]

# 4. Display and save the image
image.show()
image.save("generated_task2.png")
print("Image generated and saved successfully!")
```
